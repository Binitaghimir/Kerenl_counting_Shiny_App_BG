```python
import os
import cv2
import pandas as pd
from ultralytics import YOLO

# ================= CONFIG =================
model_path = r"C:\Users\bg61024\OneDrive - University of Georgia\Research Materials\Proposal\Machine_learning_paper\Shiny_app\Kerenl_counting_Shiny_App_BG\best.pt"

image_dir = r"C:\Users\bg61024\OneDrive - University of Georgia\Research Materials\Proposal\Machine_learning_paper\Shiny_app\Try_image"

output_dir = r"C:\Users\bg61024\OneDrive - University of Georgia\Research Materials\Proposal\Machine_learning_paper\Shiny_app\Try_image\output2"
os.makedirs(output_dir, exist_ok=True)

vis_dir = os.path.join(output_dir, "visual_predictions")
os.makedirs(vis_dir, exist_ok=True)

CONF = 0.467
NMS_IOU = 0.70
IMG_SIZE = 640
MAX_DET = 3000

MULTIPLY_BY_2 = False   # set True only if image shows one side but you want total kernel estimate


# ================= LOAD MODEL =================
model = YOLO(model_path)

results_list = []

# ================= PREDICT IMAGES =================
for img_name in os.listdir(image_dir):

    if not img_name.lower().endswith((".jpg", ".jpeg", ".png")):
        continue

    img_path = os.path.join(image_dir, img_name)
    img = cv2.imread(img_path)

    if img is None:
        print(f"Could not read image: {img_name}")
        continue

    result = model.predict(
        source=img,
        imgsz=IMG_SIZE,
        conf=CONF,
        iou=NMS_IOU,
        max_det=MAX_DET,
        verbose=False
    )[0]

    boxes = result.boxes.xyxy.cpu().numpy() if result.boxes is not None else []

    predicted_count = len(boxes)

    if MULTIPLY_BY_2:
        predicted_count = predicted_count * 2

    # draw center dots
    vis_img = img.copy()

    for x1, y1, x2, y2 in boxes:
        cx = int((x1 + x2) / 2)
        cy = int((y1 + y2) / 2)

        cv2.circle(vis_img, (cx, cy), 6, (0, 0, 255), -1)

    cv2.putText(
        vis_img,
        f"Predicted kernels: {predicted_count}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1.2,
        (255, 255, 255),
        3,
        cv2.LINE_AA
    )

    out_img_path = os.path.join(vis_dir, img_name)
    cv2.imwrite(out_img_path, vis_img)

    results_list.append({
        "Image_Name": img_name,
        "Predicted_Kernel_Count": predicted_count,
        "Confidence": CONF,
        "NMS_IOU": NMS_IOU,
        "Visual_Output": out_img_path
    })

    print(f"{img_name}: {predicted_count} kernels")


# ================= SAVE EXCEL =================
df = pd.DataFrame(results_list)

excel_path = os.path.join(output_dir, "predicted_kernel_counts.xlsx")
df.to_excel(excel_path, index=False)

print("\nDone.")
print("Excel saved at:", excel_path)
print("Visual predictions saved at:", vis_dir)
```
